```python
from transformers import T5Tokenizer, T5ForConditionalGeneration

# 1️⃣ Charger le modèle T5-base
model_name = "t5-base"
tokenizer = T5Tokenizer.from_pretrained(model_name)
model = T5ForConditionalGeneration.from_pretrained(model_name)

# 2️⃣ Texte brut
texte_brut = """
5/01/2016 13:51
Montrer qu'il fait preuve de ruse pour déjouer les gardiens.
Il pourrait entendre le début du show.
IDÉE : Qu'on ne sache pas vraiment ce que Damian fait là. On pourrait
penser qu'il est en danger de mort, que c'est la guerre de l'autre côté du mur.
Bref, écrire ça pour que le lecteur soit surpris d'apprendre que
c'est bien du théâtre, et non une vraie guerre, qui se passe de l'autre côté du mur.

12/01/2016 13:10
Damian s'est attaché un foulard indigo autour du cou, pour faire
comme le Grand Vaöstel.
"""

# 3️⃣ Entités NER détectées
ner_entities = [
    ("5/01/2016 13:51", "DATE"),
    ("12/01/2016 13:10", "DATE"),
    ("Damian", "PER"),
    ("Grand Vaöstel", "PER"),
    ("mur", "LOC"),        # optionnel
    ("théâtre", "MISC"),   # optionnel
    ("show", "MISC")       # optionnel
]

# 4️⃣ Reconstruire un texte contextuel lisible à partir des entités et notes
texte_contextuel = texte_brut.strip()
# Optionnel : mettre en valeur les PER et autres entités
for entity, label in ner_entities:
    if label in ["PER", "LOC", "DATE", "ORG", "MISC"]:
        texte_contextuel = texte_contextuel.replace(entity, f"{entity}")

# 5️⃣ Préparer l'entrée pour T5
input_text = "summarize: " + texte_contextuel
print(input_text)
inputs = tokenizer.encode(input_text, return_tensors="pt", max_length=512, truncation=True)

# 6️⃣ Générer le résumé
summary_ids = model.generate(
    inputs,
    max_length=120,  # ajustable selon besoin
    min_length=50,
    num_beams=4,
    early_stopping=True
)

# 7️⃣ Décoder et afficher le résumé
summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
# print("Résumé généré :\n", summary)
```
